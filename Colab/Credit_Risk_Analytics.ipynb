```python
cols = [
    'person_income',
    'person_age',
    'person_emp_length'
]

for col in cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count_ = ((df[col] < lower) |
             (df[col] > upper)).sum()



    Pct = (count_/len(df)) * 100
    print(col, ":", count_)
    print("Outliers Pct % :",Pct)


```

**Handling The Outlier**

```python
cols=["person_income","person_age","person_emp_length"]
Q1 = df[cols].quantile(0.25)
Q3= df[cols].quantile(0.75)
IQR = Q3-Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR


df = df[((df[cols] >= lower) & (df[cols] <= upper)).all(axis=1)]
df
```

```python
cols = [
    'person_income',
    'person_age',
    'person_emp_length'
]
for col in cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count_ = ((df[col] < lower) |
             (df[col] > upper)).sum()

    PctOutlier = (count_ / len(df)) * 100

    print(col, ":", count_)
    print(f"After Treatment PCT :",PctOutlier )

```

**Does employment or home Ownership type make a difference?**


```python
employment_status = (
    df.groupby(['employment_type', 'loan_status'])
      .size()
      .unstack(fill_value=0)
)

employment_status.columns = ['Non Defaulters', 'Defaulters']

employment_status['Total'] = (
    employment_status['Non Defaulters'] +
    employment_status['Defaulters']
)

# Percentage of total borrowers
# employment_status['Percentage'] = (
#     employment_status['Total'] /
#     employment_status['Total'].sum() * 100
# ).round(2)
employment_status['Default Rate %'] = (
    employment_status['Defaulters'] /
    employment_status['Total'] * 100
).round(2)
print(employment_status)
```

```python
sns.countplot(
    data=df,
    x='employment_type',
    hue='Default_status'
)
plt.show()
```

**Observation:**

* Default rates are very similar across all employment categories.
* The difference between the highest and lowest default rate is less than 1 percentage point.

This indicates that employment type has little influence on loan default behavior in this dataset.

**Conclusion:**

Employment type does not appear to be a strong predictor of loan default, as borrowers across all employment categories exhibit nearly identical default rates.

**Are there clear differences between borrowers in the USA, UK, and Canada?**

```python
country_summary = (
    df.groupby('country')
      .agg(
          Total=('loan_status', 'count'),

      )
)


print(country_summary)
```

```python

country_summary = country_summary.reset_index()

plt.figure(figsize=(10,5))

sns.barplot(
    data=country_summary,
    x='country',
    y='Total'
)

plt.title('Total Loans by Country')
plt.xlabel('Country')
plt.ylabel('Number of Loans')
plt.xticks(rotation=45)

plt.show()
```

 **The loan distribution is very balanced across all three countries.**
* The UK has the highest number of loans (9,752), but only by a small margin.
* The USA follows closely with 9,703 loans.
* Canada has the lowest count (9,649 loans), though the difference is minimal.

**Key Observations**
1. Borrowers are evenly distributed across countries
The number of loan applications is relatively similar across the USA, UK, and Canada.
No single country dominates the borrower population.

2. Balanced geographic representation
Each country contributes a comparable share of total borrowers.
This indicates that the dataset is geographically balanced and not biased toward a specific country.

3. Stable portfolio diversification
Since borrowers are distributed across multiple countries, the loan portfolio is not concentrated in one region.
Geographic diversification helps reduce country-specific risk exposure.

4. Comparable market participation
Borrowing activity appears consistent across all three countries, suggesting similar levels of demand for loans within the dataset.

**Conclusion**

The borrower base is well distributed across the USA, UK, and Canada, with no country showing a disproportionate share of loan applications. This balanced geographic distribution indicates a diversified loan portfolio and reduces dependency on any single market.

# **Executive Summary**

**Descriptive Analysis**

The portfolio is large and diverse, but the headline concern is a **22.5%** default rate. Grade A and B borrowers dominate by count (18,956 combined), but Grades D through G are catastrophically risky. **Grade G alone defaults 98.21%** of the time. Renters make up **52%** of borrowers, and loan purposes skew toward financial distress signals like medical debt and debt consolidation.

**Diagnostic Analysis**

**Four compounding root causes drive defaults**
* DTI > 50% → default rate of 56.66% (quintuples vs. lowest band)
* LTI > 30% → default rate of 67.45% (6x the low-burden cohort)

* Income < 30K → default rate of 47.27% (4.4x vs. 75K+ earners)
* Grade D–G → default rates of 60–98%, validating the internal grade ladder as a reliable but currently under-acted-on risk signal

**Predictive Analysis**

Without intervention, any macroeconomic shock (rate hikes, wage pressure) will push at-risk exposure from **69M** toward **82-85M**. A borrower combining low income, high DTI, high LTI, and a weak loan grade faces an estimated default probability exceeding 85%.

**Prescriptive Analysis**

Three-horizon action plan:

* **Immediate (0–3 months):** Hard-cap LTI at **25%**, decline DTI > 45%, suspend Grade F/G originations, triage high-risk existing accounts

* **Medium-term (3–12 months):** Mandatory income verification for **sub 35K**, risk-adjusted interest pricing by grade, affordability assessments for medical/debt consolidation loans
* **Long-term (12–24 months):** Real-time early warning system, portfolio rebalancing toward Grade A–B and 50K+ income segments, secured lending for high-risk but viable borrowers

**Estimated impact:** Targeting a reduction from 22.5% toward 15–17% default rate within 18 months, protecting approximately $15–20M in at-risk capital annually.
